```python
import pandas as pd
from openbabel import pybel
```

# Set parameters

```python
smiles_csv_path = "../../../pre-trained/prepared_passed.csv"

out_smi_path = "navitoclax_analogues.smi"
out_smi_protonated_path = "navitoclax_analogues_protonated.smi"
out_csv_protonated_path = "../../../pre-trained/protonated_prepared_passed.csv"
```

# Run protonation
This code protonates SMILES from a CSV file using OpenBabel at pH 7.4.

```python
pd.read_csv(smiles_csv_path)[["SMILES", "Mol ID"]].to_csv(out_smi_path, sep="\t", header=False, index=False)

output = pybel.Outputfile("smi", out_smi_protonated_path)

for mol in pybel.readfile('smi', out_smi_path):
    mol.OBMol.AddHydrogens(False, True, 7.4)
    output.write(mol)
    
output.close()

df = pd.read_csv(out_smi_protonated_path, sep="\t", names=["SMILES", "Mol ID"])
df[["Mol ID", "SMILES"]].to_csv(out_csv_protonated_path, index=False)
```
