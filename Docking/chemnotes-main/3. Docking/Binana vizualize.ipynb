```python
import pandas as pd
from chemplus import binana
pd.set_option("display.max_colwidth", None)
```

# Set parameters

```python
protein_pdbqt = "test/hr1(2).pdbqt"
ligs_pdbqt_top_model_dir = "test/top_model"
ligs_docked_sdf = "test/sm_compound_2runs_docked.sdf.gz"

pymol_out_session = "test/binana.pse"
binana_out_dir = "test/binana_out"
binana_analysis_csv = binana_out_dir + "/binana_results.csv"

lig_names_list = list(pd.read_csv("test/top10.csv")["Name"])
```

# Run binana and vizualization

```python
binana.binana_vizualize(protein_pdbqt, ligs_pdbqt_top_model_dir, ligs_docked_sdf, binana_out_dir, 
                        binana_analysis_csv, pymol_out_session, lig_names_list=lig_names_list)

df_binana = pd.read_csv(binana_analysis_csv)

columns_to_show = []
for column in df_binana.columns:
    if not df_binana[column].isna().all():
        columns_to_show.append(column)

from IPython.display import HTML
HTML(df_binana[columns_to_show].fillna("-").to_html(index=False))
```
